#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.


```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Set pandas option to display all columns
pd.set_option('display.max_columns', None)

# Display the first few rows of the dataset
df.head()
```

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
# Check missing values
df[['Employment', 'JobSat', 'RemoteWork']].isnull().sum()
```

```python
# Impute RemoteWork using mode
df['RemoteWork'] = df['RemoteWork'].fillna(
    df['RemoteWork'].mode()[0]
)

# Impute JobSat using median
df['JobSat'] = df['JobSat'].fillna(
    df['JobSat'].median()
)
```

```python
# Verify results
df[['Employment', 'JobSat', 'RemoteWork']].isnull().sum()
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
# Convert text values into numeric values first.

df['YearsCodePro_clean'] = df['YearsCodePro'].replace({
    'Less than 1 year': 0.5,
    'More than 50 years': 50
})

df['YearsCodePro_clean'] = pd.to_numeric(
    df['YearsCodePro_clean'],
    errors='coerce'
)

# Create experience ranges
bins = [0, 5, 10, 20, 100]
labels = ['0-5 years', '5-10 years', '10-20 years', '>20 years']

df['ExperienceRange'] = pd.cut(
    df['YearsCodePro_clean'],
    bins=bins,
    labels=labels
)

# Calculate median Job Satisfaction
jobsat_by_exp = df.groupby('ExperienceRange')['JobSat'].median()

print(jobsat_by_exp)
```

```python
# Visualize with a bar plot
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.barplot(
    x=jobsat_by_exp.index,
    y=jobsat_by_exp.values
)

plt.title('Median Job Satisfaction by Experience Range')
plt.xlabel('Experience Range')
plt.ylabel('Median Job Satisfaction')

plt.show()
```

After grouping respondents into experience ranges based on YearsCodePro, the median JobSat score was calculated for each group. All experience ranges showed a median job satisfaction score of 7.0, indicating relatively consistent job satisfaction levels across different professional experience levels.

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
plt.figure(figsize=(8,5))

sns.countplot(x='JobSat', data=df)

plt.title('Distribution of Job Satisfaction')
plt.xlabel('Job Satisfaction Score')
plt.ylabel('Count')

plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
# Remote Work Distribution (Count Plot)
plt.figure(figsize=(8,5))

sns.countplot(x='RemoteWork', data=df)

plt.title('Remote Work Preferences')
plt.xlabel('Work Type')
plt.ylabel('Count')

plt.xticks(rotation=30)

plt.show()
```

```python
# Cross-tab: Remote Work vs Employment Type: This shows how job roles relate to remote work preferences.

remote_by_employment = pd.crosstab(
    df['Employment'],
    df['RemoteWork']
)

print(remote_by_employment)
```

```python
# Visual heatmap (recommended for clarity)
plt.figure(figsize=(12,6))

sns.heatmap(remote_by_employment, annot=True, fmt='d', cmap='Blues')

plt.title('Remote Work by Employment Type')

plt.show()
```

Independent contractors demonstrated the highest preference for remote work, while students were primarily associated with hybrid arrangements. Overall, in-person work was less common compared to hybrid and remote models, reflecting current trends in the technology sector.

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
# step 1: Select countries/regions: Choose a few major countries for comparison.

selected_countries = [
    'United States of America',
    'India',
    'Germany',
    'Canada'
]

df_region = df[df['Country'].isin(selected_countries)]

# Step 2: Split programming languages: The LanguageHaveWorkedWith column contains multiple languages separated by ;.

# We split them into individual rows using explode().

languages = df_region[['Country', 'LanguageHaveWorkedWith']].dropna()

languages['LanguageHaveWorkedWith'] = (
    languages['LanguageHaveWorkedWith'].str.split(';')
)

languages = languages.explode('LanguageHaveWorkedWith')

# Step 3: Count language popularity by country
language_counts = (
    languages.groupby(['Country', 'LanguageHaveWorkedWith'])
    .size()
    .reset_index(name='Count')
)

# Step 4: Get top languages
top_languages = (
    language_counts.sort_values(['Country', 'Count'], ascending=[True, False])
    .groupby('Country')
    .head(5)
)

print(top_languages)

# Step 5: Visualize with bar plot
plt.figure(figsize=(14,6))

sns.barplot(
    data=top_languages,
    x='LanguageHaveWorkedWith',
    y='Count',
    hue='Country'
)

plt.title('Top Programming Languages by Country')
plt.xlabel('Programming Language')
plt.ylabel('Count')

plt.xticks(rotation=45)

plt.show()
```

```python
heatmap_data = language_counts.pivot_table(
    index='Country',
    columns='LanguageHaveWorkedWith',
    values='Count',
    fill_value=0
)

# Keep top overall languages only
top_cols = heatmap_data.sum().sort_values(ascending=False).head(10).index
heatmap_data = heatmap_data[top_cols]

plt.figure(figsize=(12,6))

sns.heatmap(heatmap_data, cmap='Blues')

plt.title('Programming Language Popularity by Country')

plt.show()
```

Programming language trends were analyzed across selected countries using the LanguageHaveWorkedWith column. JavaScript emerged as the most commonly used language in all analyzed regions, followed closely by Python and SQL. India showed comparatively stronger Java usage, while Bash/Shell tools were highly common in Germany and the United States. These findings highlight globally shared technology trends alongside regional differences in language preferences.

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
# Step 1: Clean YearsCodePro: Convert text values into numeric values.

df['YearsCodePro_clean'] = df['YearsCodePro'].replace({
    'Less than 1 year': 0.5,
    'More than 50 years': 50
})

df['YearsCodePro_clean'] = pd.to_numeric(
    df['YearsCodePro_clean'],
    errors='coerce'
)

# Step 2: Remove rows with missing values
scatter_data = df[['YearsCodePro_clean', 'JobSatPoints_1']].dropna()

# Step 3: Create scatter plot
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=scatter_data,
    x='YearsCodePro_clean',
    y='JobSatPoints_1'
)

plt.title('Experience vs Job Satisfaction')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Job Satisfaction Points')

plt.show()
```

```python
# Add regression trend line: This makes correlation easier to observe.

plt.figure(figsize=(8,5))

sns.regplot(
    data=scatter_data,
    x='YearsCodePro_clean',
    y='JobSatPoints_1',
    scatter_kws={'alpha':0.5}
)

plt.title('Experience vs Job Satisfaction')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Job Satisfaction Points')

plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
# Step 1: Clean Employment column. Since Employment contains multiple combined values, use the primary category only.

df['Employment_clean'] = df['Employment'].str.split(';').str[0]

# Step 2: Create cross-tabulation
education_employment = pd.crosstab(
    df['EdLevel'],
    df['Employment_clean']
)

print(education_employment)

# Step 3: Visualize using a heatmap
plt.figure(figsize=(12,6))

sns.heatmap(
    education_employment,
    cmap='Blues'
)

plt.title('Education Level vs Employment Type')

plt.xlabel('Employment Type')
plt.ylabel('Education Level')

plt.show()
```
