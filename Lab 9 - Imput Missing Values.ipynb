### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>
The functions below will download the dataset into your browser:
</p>


```python
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
df.duplicated().sum()
```

```python
duplicates = df[df.duplicated()]
duplicates.head()
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
# Remove duplicates
df_clean = df.drop_duplicates()
```

```python
# Verify removal
df_clean.duplicated().sum()
```

```python
print("Original dataset shape:", df.shape)
print("After removing duplicates:", df_clean.shape)
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
#Find Missing Values for All Columns
missing_values = df_clean.isnull().sum()
print(missing_values)
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
df_clean['RemoteWork'].isnull().sum()
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
df_clean['RemoteWork'].value_counts()
```

The dataset has 3 main work arrangement types

The most common category is:

👉 Hybrid (some remote, some in-person)

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
df_clean['RemoteWork'].mode()
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
mode_value = df_clean['RemoteWork'].mode()[0]
df_clean['RemoteWork'] = df_clean['RemoteWork'].fillna(mode_value)

#Verify imputation worked
df_clean['RemoteWork'].isnull().sum()
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
df_clean[['CompTotal', 'ConvertedCompYearly']].describe()
```

The compensation-related columns analyzed were CompTotal and ConvertedCompYearly. The CompTotal column contains extreme outliers and unrealistic values, making it unsuitable for analysis without cleaning. In contrast, ConvertedCompYearly provides a more standardized measure of annual compensation. Its distribution is highly right-skewed, with most values concentrated between the lower and mid-range salaries, while a few extreme values significantly increase the maximum and mean.
